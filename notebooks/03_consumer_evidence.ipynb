## Published front-of-pack label studies (observed)

```python
ev = pd.read_csv(ROOT / 'data' / 'consumer_evidence' / 'fop_label_studies.csv')
ev[['study','year','country','label_type','main_finding']]
```

## What the literature consistently shows
- Interpretive / colour / warning labels shift purchases away from flagged products (clearest case: Chile's warning labels).
- The effect is **partial**, not wholesale (Nutri-Score experiments).
- Labels also push **reformulation** (meta-analytic evidence).
- India's own scheme is the **star-based INR**, not Nutri-Score.

## Scenario assumptions (this project — NOT empirical predictions)
Base switch rates of the D/E assortment, chosen within the evidence range and scaled by category consumer sensitivity.

```python
scen = pd.read_csv(FINAL / 'scenario_results.csv')
assumed = (scen[['scenario','effective_switch_rate']]
           .groupby('scenario')['effective_switch_rate']
           .agg(['min','mean','max']).round(3))
print('base rates: Low 4% / Moderate 10% / High 20% (assumptions)')
assumed
```
